```python
# Step 1: Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# Configure visualizations
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Locate project root and dataset
PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "raw" / "WA_Fn-UseC_-HR-Employee-Attrition.csv"

# Load dataset
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# Display first five records
display(df.head())

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
display(df.describe())

# Categorical column summary
print("\nCategorical Feature Summary:")
display(df.describe(include="str"))
```

```python
# Step 4: Attrition distribution

# Calculate counts and percentages
attrition_counts = df["Attrition"].value_counts()
attrition_percentages = df["Attrition"].value_counts(normalize=True) * 100

print("Attrition Distribution:")
for category in attrition_counts.index:
    print(
        f"{category}: {attrition_counts[category]} employees "
        f"({attrition_percentages[category]:.2f}%)"
    )

# Plot distribution
plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df,
    x="Attrition",
    order=["No", "Yes"]
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

# Add count labels
for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()

# Save graph
FIGURE_DIR = PROJECT_ROOT / "reports" / "figures"
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

plt.savefig(
    FIGURE_DIR / "01_attrition_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Step 5: Attrition by Age Group

# Create age groups
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-60"]
)

# Calculate attrition counts
age_attrition = pd.crosstab(
    df["AgeGroup"],
    df["Attrition"]
)

print("Attrition Count by Age Group:")
display(age_attrition)

# Calculate attrition percentage within each age group
age_attrition_percent = pd.crosstab(
    df["AgeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

print("\nAttrition Percentage by Age Group:")
display(age_attrition_percent.round(2))

# Plot attrition percentage
age_attrition_percent["Yes"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Attrition Percentage by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Employees Who Left (%)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    FIGURE_DIR / "02_attrition_by_age.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Step 10: Attrition by Monthly Income

# Divide employees into four income groups
df["IncomeGroup"] = pd.qcut(
    df["MonthlyIncome"],
    q=4,
    labels=["Low", "Lower-Middle", "Upper-Middle", "High"]
)

# Calculate attrition counts
income_attrition = pd.crosstab(
    df["IncomeGroup"],
    df["Attrition"]
)

print("Attrition Count by Income Group:")
display(income_attrition)

# Calculate attrition percentages
income_attrition_percent = pd.crosstab(
    df["IncomeGroup"],
    df["Attrition"],
    normalize="index"
) * 100

print("\nAttrition Percentage by Income Group:")
display(income_attrition_percent.round(2))

# Plot
income_attrition_percent["Yes"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Attrition Percentage by Monthly Income Group")
plt.xlabel("Monthly Income Group")
plt.ylabel("Employees Who Left (%)")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig(
    FIGURE_DIR / "07_attrition_by_income.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# Step 13: Correlation Analysis

# 1. Encode Attrition
df["AttritionBinary"] = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})

# 2. Define columns to exclude
columns_to_exclude = [
    "EmployeeCount",
    "StandardHours",
    "EmployeeNumber"
]

# 3. Select numerical columns and remove excluded columns
numerical_df = df.select_dtypes(include="number").drop(
    columns=columns_to_exclude,
    errors="ignore"
)

# 4. Calculate correlations with Attrition
target_correlations = (
    numerical_df
    .corr()["AttritionBinary"]
    .drop("AttritionBinary")
    .sort_values()
)

print("Correlation of Numerical Features with Attrition:")
display(target_correlations.to_frame(name="Correlation").round(4))

# 5. Generate correlation matrix
correlation_matrix = numerical_df.corr()

# 6. Plot heatmap
plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.3,
    annot=False
)

plt.title("Correlation Heatmap - Numerical Features")
plt.tight_layout()

# 7. Save heatmap
plt.savefig(
    FIGURE_DIR / "10_correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nCorrelation analysis completed successfully!")
```
