```python

# Milestone 5: Data Preprocessing
# Step 1: Import libraries and load dataset

import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Locate project root
PROJECT_ROOT = Path.cwd().parent

# Dataset location
DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "raw"
    / "WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

# Load original dataset
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

# Check target distribution
print("\nOriginal Attrition Distribution:")
print(df["Attrition"].value_counts())

# Check missing values
print("\nTotal missing values:", df.isnull().sum().sum())

# Check duplicate records
print("Duplicate records:", df.duplicated().sum())
```

```python

# Step 2: Feature Selection and Target Separation

# Define columns to remove
columns_to_drop = [
    "EmployeeNumber",
    "EmployeeCount",
    "Over18",
    "StandardHours"
]

# Separate target variable
y = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})

# Create feature dataset
X = df.drop(
    columns=["Attrition"] + columns_to_drop
)

# Display results
print("Feature and Target Separation Completed!")

print("\nOriginal Dataset Shape:", df.shape)
print("Feature Matrix Shape:", X.shape)
print("Target Variable Shape:", y.shape)

print("\nRemaining Features:")
print(X.columns.tolist())

print("\nTarget Distribution:")
print(y.value_counts().sort_index())

print("\nTarget Mapping:")
print("No  = 0")
print("Yes = 1")
```

```python

# Step 3: Identify Numerical and Categorical Features

# Identify numerical columns
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

# Identify categorical columns
categorical_features = X.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

# Display results
print("Numerical Features:", len(numerical_features))
print("Categorical Features:", len(categorical_features))

print("\nNumerical Feature Columns:")
for feature in numerical_features:
    print("-", feature)

print("\nCategorical Feature Columns:")
for feature in categorical_features:
    print("-", feature)

print("\nTotal Identified Features:",
      len(numerical_features) + len(categorical_features))
```

```python

# Step 4: Create Preprocessing Pipelines

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Numerical preprocessing pipeline
numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing pipeline
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

# Combine both pipelines
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipelines created successfully!")

print("\nNumerical pipeline:")
print(numerical_transformer)

print("\nCategorical pipeline:")
print(categorical_transformer)

print("\nCombined preprocessor:")
print(preprocessor)
```

```python

# Step 5: Train-Test Split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dataset shapes
print("Train-Test Split Completed Successfully!")

print("\nTraining Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)

print("\nTraining Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)

# Check class distribution
print("\nTraining Set Attrition Distribution:")
print(y_train.value_counts())

print("\nTesting Set Attrition Distribution:")
print(y_test.value_counts())

# Check class percentages
print("\nTraining Set Attrition Percentages:")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTesting Set Attrition Percentages:")
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python

# Step 6: Fit and Apply Preprocessing Pipeline

# Fit preprocessing pipeline on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data using the fitted pipeline
X_test_processed = preprocessor.transform(X_test)

# Display transformed shapes
print("Preprocessing Completed Successfully!")

print("\nOriginal Training Shape:", X_train.shape)
print("Processed Training Shape:", X_train_processed.shape)

print("\nOriginal Testing Shape:", X_test.shape)
print("Processed Testing Shape:", X_test_processed.shape)

print("\nTraining Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)

# Verify missing values after preprocessing
print("\nMissing Values in Processed Training Data:",
      np.isnan(X_train_processed).sum())

print("Missing Values in Processed Testing Data:",
      np.isnan(X_test_processed).sum())
```

```python

# Step 7: Save Preprocessing Artifacts

import joblib

# Define output directories
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
MODEL_DIR = PROJECT_ROOT / "models"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
MODEL_DIR.mkdir(parents=True, exist_ok=True)

# Save fitted preprocessing pipeline
joblib.dump(
    preprocessor,
    MODEL_DIR / "attrition_preprocessor.joblib"
)

# Save processed feature datasets
np.save(
    PROCESSED_DIR / "X_train_processed.npy",
    X_train_processed
)

np.save(
    PROCESSED_DIR / "X_test_processed.npy",
    X_test_processed
)

# Save target datasets
np.save(
    PROCESSED_DIR / "y_train.npy",
    y_train.to_numpy()
)

np.save(
    PROCESSED_DIR / "y_test.npy",
    y_test.to_numpy()
)

print("Preprocessing artifacts saved successfully!")

print("\nSaved Files:")
print("- models/attrition_preprocessor.joblib")
print("- data/processed/X_train_processed.npy")
print("- data/processed/X_test_processed.npy")
print("- data/processed/y_train.npy")
print("- data/processed/y_test.npy")
```

```python

# Step 8: Verify Saved Preprocessing Artifacts

import joblib
import numpy as np

# Define file paths
preprocessor_path = MODEL_DIR / "attrition_preprocessor.joblib"

x_train_path = PROCESSED_DIR / "X_train_processed.npy"
x_test_path = PROCESSED_DIR / "X_test_processed.npy"

y_train_path = PROCESSED_DIR / "y_train.npy"
y_test_path = PROCESSED_DIR / "y_test.npy"

# Load saved artifacts
loaded_preprocessor = joblib.load(preprocessor_path)

loaded_X_train = np.load(x_train_path)
loaded_X_test = np.load(x_test_path)

loaded_y_train = np.load(y_train_path)
loaded_y_test = np.load(y_test_path)

# Verify shapes
print("Saved Artifacts Loaded Successfully!")

print("\nLoaded Dataset Shapes:")
print("X_train:", loaded_X_train.shape)
print("X_test:", loaded_X_test.shape)
print("y_train:", loaded_y_train.shape)
print("y_test:", loaded_y_test.shape)

# Verify consistency
assert loaded_X_train.shape == (1176, 51)
assert loaded_X_test.shape == (294, 51)
assert loaded_y_train.shape == (1176,)
assert loaded_y_test.shape == (294,)

print("\nAll shape checks passed!")

# Verify preprocessor
assert loaded_preprocessor is not None

print("Preprocessor loaded successfully!")
print("\nAll preprocessing artifacts verified successfully!")
```
